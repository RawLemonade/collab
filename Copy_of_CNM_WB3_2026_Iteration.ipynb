There are two main types of loop: a for loop and a while loop. We will explore both of these below.

Iteration isn't the only way of reusing a bit of code. We can also write our code as a function which we can call repeatedly with different arguments. ## Iteration


Now to learn how to [iterate](http://docs.python.org/tutorial/controlflow.html#for-statements) over code...

Some things to note:


*   Note the structure here. Blocks of code are identified by indentation. The `for` statement includes a `:` at the end, which is a way
 of telling Python (and us), that the code following it is part of a block.
*   Unlike when we were assigning variable names before with by typing `var = ` With a for loop, we assign the variable name after the `for` command. `for var in list:` where `var` is a variable name of your choosing and `list` is the object you want to iterate over.

````python
#Python has a build-in function "range()"
#It is one of the quickest ways for us to set a loop to iterate a specified number of times
#It simply creats a iterable range object, starting from zero and adding one as many times as the argument specifies.
#Let's take a look at range
print("output of range(10):")
print(range(10))

#note how Python will start from 0 and finish at 9 (NOT 10!!!). There are 10 values in total.


print("")
print("A simple loop:")
#This is the syntax of a simple "for loop". For i in range(3) will loop over the code listed below 3 times. Each time the loop is repeated,
#i is incremented by 1.
for i in range(3):
	print(i)
#Note the structure here. Blocks of code are identified by the indentation. The for statement includes a ":" at the end, which is a way
#of telling Python (and us), that the below code is part of a block. Python will loop through all code below indented one level more than
#the for loop definition.

print("")
print("The importance of correct spacing:")
for i in range(5):
	print("this will be looped 5 times")
print("this will not be looped as it is not indented one tab from the for loop")

print("")
print("range can be implemented like arange too with a start, stop and increment:")
print("this will start at 3 and end at 5")
for i in range(3, 6, 1):
	print(i)

print("")
print("range can also count down:")
print("this will count down")
for i in range(10, -1, -1): # we have to set the end to -1 if we want to get all the way to zero
	print(i)

print("")
print("using other means to seed a for loop:")
#it is also pretty darn similar to create loops based on other factors. Here is an example which loops for the number of characters in a string.
a_string = "supercalafragalisticexpialadoshus"



print("looping over a string:")
for letter in a_string:
	print(letter)

#the len() function prints the length of a variable
print("The loop did this many iterations based on the length of the string:")
print(len(a_string))

print("")
print("Here's a nested loop:")
# A LOOP IN A LOOP!!
#Nested loops are often very useful...

z = 0

for i in range(2):
	for j in range(3):
		z += 1			#you've not seen "+=" yet. It is simply shorthand for saying "variable = variable + ..."
						    #so here "z += 1" is the same as "z = z + 1" i.e incramentally increasing the value of your variable.


print(z)			#The answer, 6 of course! z is incremented 3 times, twice!



print("")
print("Getting out of there...:" )

#Sometimes, we set a max limit for a loop, but place a condition that, once met, escapes the loop early.
for number in range(1000):
	print(number)
	if number == 3: #this is a comparison, explained in a later lesson
		break	#the break statement is what tells Python to stop looping. This will be called when i is 3.

print("Got out of there!")


#In addition to a for loop, while loops provide another means to repeat code. While loops are useful when you don't know in advance
#how many times you'll need to loop (for example when waiting for user interaction).
#for now, here's a simple demonstration
print("")
print("This was looped within a 'while':")

ticker = 1
while ticker <= 5:    #<= means less than or equal to! The function will continue to loop until i > 5!
	print(ticker)
	ticker += 1 # incramentally increase the value of 'ticker'


#Some programs have statements which loop forever! use ctrl+c to break it (or interrupt the kernel if in Jupyter)
#while True:
	#print "To infinity, and beyond!"
````

## Conditions

http://docs.python.org/tutorial/controlflow.html#if-statements

Most programs have conditions in order to decide an appropriate means to process data going forward...

Some things to note:


*   Note the same structure used here as with for and while loops. Special python word, info, colon then intended code block.


````python
#setting up some variables
x = 1
y = 2
z = 1
j = 0
k = -1

#remember these boolean types from lab 1?
varA = True
varB = False


print("")
print("Does x = 1:")

if x == 1:				#the "if" statement evaluates an expression - if the expression evaluates to a 'TRUE' (or a number that isn't 0),
						      #then the below code block is executed
	print("x = 1")
#note how we DO NOT say if x = 1. We use two '=' signs. = sets the value of something. i.e. x = 1. == is used to compare,
#and is what we should use here.


print("")
print("Multiple IFs, and default code:")

#here we compare the value of x to 2. If it doesn't match, the program then compares the value to 3. if that doesn't match, the code
#block under the 'else' statement is executed. "else" is used to specify what to execute if no conditions are met

if x == 2:
	print("x = 2")
elif x == 3:
	print("x = 3")
else:
	print("x was not 2 or 3")


print("")
print("Comparing two values:")

#failry obvious what this does...
if x == z:
	print("x and z are the same")
# Note that you don't have to close your if statement with an else statement, but it is good practise to do so.


print("")
print("The pass statement is used")

#There MUST be code indented after a ':'. If you have a condition, but don't have anything to run if the condition is met, then use "pass".
#Pass does nothing :)
if x == z:
	pass


print("")
print("It works with all sorts of compairsons...:")
#obvious...
if x < y:
	print("x is less than y")


print("")
print("unusual conditions:")
#as mentioned before, if the result of an expression is TRUE or > 0, code is executed. Here we are evaluating only y, which is 2, which is not 0.
#code is therefore executed...
if y:
	print("y is not 0")

if j:								# j is zero, so the code under "else" will be executed
	print("j is not 0")
else:
	print("j must be 0 or null")

if k:
	print("k is not zero")		#as long as it's not 0! k is -1, which is not 0!



print("")
print("evaluating booleans:")
if varA:							# if true...
	print("TRUE")

print("")
print("A double negative...:")
if not varB:						#if NOT false. A double negative for you :)
	print("TRUE")

````

## Dictionaries

Here is another python object that might be useful. Dictionaries are like lists, but instead of indexing an item by a number (e.g. `list[i]`), you index values by keys. They're like a simple table. You have a "key" and a corresponding "value". We are looking at them here because they are another useful way to store information that you can then iterate over.

````python
# http://docs.python.org/tutorial/datastructures.html#dictionaries


print("")
print("Here's a dictionary:")
#You create a dictionary with curly brackets {} and put in key value pairs as key : value, and separate pairs with commas
#"Name" has value "Liz". "Location" has value "Manchester"
information = {"Name" : "Liz", "Location": "Manchester"}

#Here's what it looks like printed
print(information)

print("")
print("Here's just the value of 'Name':")

#to select a value you reference it by the "key" in square brackets (like indexing a list or array).
print(information["Name"])


print("")
print("Changed a value...")

#Here's how you change a value.
information["Location"] = "University"

print(information)


print("")
print("Added a value...")

#Here's how to add a value. The value doesn't have to be a single item, you can store anything- a number, a string, a list, even another dictionary!
information["Occupation"] = ["Researcher", "Lecturer"]

print(information)


print("")
print("Removed a value..")

#...and to remove...
del information["Occupation"]

print(information)


print("")
print("Looping over a dictionary...")

#Looping is a little more complicated. dictionary.items() returns 2 vars, so we need to specify 2 variables in the
#for loop so as to access both keys and values.
for key, value in information.items():
	print("The key is: " + str(key) + ", and the value is: " + value)
````

````python
import time;	#Like the previous lesson, we require additional functionality not enabled by default in Python.
				#The time module will allow us to get the current system time.


#To open a file we use the "open()" function. This function requires 2 parameters:
#1: The name of the file you want to open. If you haven't specified the full path of the file (e.g. c:\somefolder\somefile.py), it will
#look in from the location where the script was called (careful, this is not always where the script lies).
#2: We specify how we want to open the file. 'r' is for read access, 'r+' is for read and write access, 'w' is for write access, and 'a' is for
#write access but to "append" and not overwrite.
f = open("open_me.txt", 'r')
#You may have noticed that we have actually created an object here. f is an object of class "open".


print("")
print("Reading all at once:")
fileContents = f.read(1024)		#the .read() method of the open() class allows us to read the contents of the file into a variable.
								#the number specified is the max number of bytes to read. If you don't specify a number, it will read the entire
								#file, which is often fine, but can be dangerous if the file is huge, for example, 4GB in size!!!

print(fileContents)		#print the file contents to the screen

#f.close()						  #Here we close the "handle" to the file.


								      #It's often nicer, easier and more efficient to read a file line by line...
print("")
print("Reading line by line:")
f2 = open("open_me.txt", 'r')	#same syntax to open the file

lineNumber = 1					#a ticker
for line in f2:					#for each line in the file object
	print("line " + str(lineNumber) + " = " + line.rstrip())		#print the line number and contents.
																#Here the rstrip() method has been called so that we don't also print the
																#newline character at the end of each line.
	lineNumber += 1				#increase ticker

f2.close()						  #close handle


#And here's how we write to a file
print("")
print("Writing to file...")

f3 = open("open_me.txt", 'w')	#open the file for writing...

theTime = time.asctime()   #don't worry about this. I just wanted to make sure the data to be written changes each time the
													 #script is executed. Can you figure out what it is doing?

f3.write(theTime)					 #we use the write() method to write stuff

f3.close()							   #close the file

print("")
print("Sleeping for 3 seconds")

time.sleep(3)						#here i use the sleep() function imported from the time module to pause execution. Again not important
								      	#to this lesson I just wanted to make sure each line was different


#Last example, this time we write to the file again, but append, instead of overwrite.
print("")
print("Appending to file...")

f4 = open("open_me.txt", 'a')

theTime = time.asctime()

f4.write("\n" + theTime)			#we make sure that we're writing on a newline by "\n".


fileContents = f.read(1024)
print(fileContents)
f.close()
f4.close()
````

There are also other modules that we can use to read numerical data into python in only a line or two of code. We often have data stored in text or CSV files. A CSV (comma-separated values) file is a simple text file that stores tabular data in a structured way, with each line representing a row and values within each row separated by commas. NumPy provides functions to read data directly into arrays for analysis.

The two most common functions are:

* **`np.loadtxt()`**
  Reads data from a text file (numbers separated by spaces, commas, or tabs).

* **`np.genfromtxt()`**
  More flexible — can handle missing values and headers.


Suppose we have a file called **`loads.csv`** with the following data:

```
10,15,20
25,30,35

```

Create this file now and upload it to colab.

We can load it into NumPy like this:


````python
# You will need to create a file called 'loads.csv' upload it to colab before this codeblock will work


# Load CSV file (comma-separated)
data = np.loadtxt("loads.csv", delimiter=",")

print("Data array:")
print(data)

# Indexing into the array
print("\nFirst row:", data[0])
print("First column:", data[:,0])

# Basic stats
print("\nMean load:", np.mean(data))
print("Maximum load:", np.max(data))


# If a file has missing entries, `np.genfromtxt()` is safer:


# Example with missing values (empty cells)
data_with_missing = np.genfromtxt("loads_with_missing.csv", delimiter=",", filling_values=0)

print("Data with missing values filled as 0:")
print(data_with_missing)


````
